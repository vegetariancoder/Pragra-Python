### Classification in Machine Learning: An Introduction
#### What is Supervised, Unsupervised and Reinforced Learning

#### _Supervised Learning_

- Concept: Supervised learning is a type of machine learning where the model learns from a labeled dataset. This means that for each input data point, there's a corresponding "correct" output or label.

##### How it works:

- Training: The model is fed a dataset containing input features and their corresponding known labels.
- Learning: The model analyzes the patterns and relationships between the input features and the labels.
- Prediction: Once trained, the model can predict labels for new, unlabeled data.

Examples:
- Spam Email Classification:
1. Input Data: Emails (features like sender, subject, content, keywords).
2. Labels: "Spam" or "Not Spam" (pre-labeled by humans).
3. Goal: Train a model to classify new incoming emails as spam or not spam.

#### _Unsupervised Learning_

- Concept: Unsupervised learning deals with unlabeled data. In this scenario, the model tries to find hidden patterns, structures, or relationships within the data on its own, without any prior knowledge of the output. It's like learning by observation, without a teacher.

##### How it works:

- Training: The model is given a dataset with only input features, no labels.
- Discovery: The model uses algorithms to identify inherent groupings, anomalies, or underlying structures in the data.
- Application: The discovered patterns can be used for various tasks like data compression, anomaly detection, or customer segmentation.


Examples:
- Customer Segmentation (Clustering):
1. Input Data: Customer demographics, purchase history, browsing behavior (no predefined customer segments).
2. Goal: Group similar customers together based on their attributes, allowing businesses to tailor marketing strategies to each segment.

#### Reinforcement Learning

- Concept: Reinforcement learning involves an "agent" that learns to make decisions by interacting with an "environment." The agent receives "rewards" for desirable actions and "penalties" for undesirable ones.

How it works:

- Agent & Environment: An agent performs actions within an environment.
- State: The environment changes its state in response to the agent's actions.
- Reward: The agent receives a numerical reward (positive or negative) based on the outcome of its action.
- Policy: The agent learns a policy that maps states to actions, aiming to maximize future rewards.

Examples:

Game Playing AI:
1. Agent: The AI program playing a game (e.g., Chess, Go, video games).
2. Environment: The game board/state.
3. Actions: Moving pieces, pressing buttons.
4. Rewards: Points for winning, penalties for losing or making bad moves.
5. Goal: Learn to play optimally and win games.

### Classification

- In classification, the model is fully trained using the training data, and then it is evaluated on test data before being used to perform prediction on new unseen data.

### Lazy Learners Vs. Eager Learners

There are two types of learners in machine learning classification: lazy and eager learners.

- `Eager learners` are machine learning algorithms that first build a model from the training dataset before making any prediction on future datasets. They spend more time during the training process because of their eagerness to have a better generalization during the training from learning the weights, but they require less time to make predictions.

- Most machine learning algorithms are eager learners, and below are some examples:

1. Logistic Regression.
2. Support Vector Machine.
3. Decision Trees.
4. Artificial Neural Networks.


- `Lazy learners or instance-based learners`, on the other hand, do not create any model immediately from the training data, and this is where the lazy aspect comes from. They just memorize the training data, and each time there is a need to make a prediction, they search for the nearest neighbor from the whole training data, which makes them very slow during prediction. Some examples of this kind are:

1. K-Nearest Neighbor.
2. Case-based reasoning.
