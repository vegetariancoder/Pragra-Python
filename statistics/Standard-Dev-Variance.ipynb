### Using Statistics

```python
import statistics

age_1 = [2,2,4,4]
age_2 = [1,1,5,5]

var = statistics.pvariance(age_1)
std = statistics.pstdev(age_1)

print("var :", var )
print("std :", std )
```

### Using Numpy

```python
import numpy as np

age_1 = np.array([2,2,4,4])
age_2 = np.array([1,1,5,5])

var = np.var(age_2)
std = np.std(age_2)

print("var :", var )
print("std :", std )
```
