# 📊 Introduction to Statistics

## 1. What is Statistics?
- **Statistics** is the branch of mathematics that deals with **collecting, organizing, analyzing, and interpreting data**.
- It helps us make decisions or predictions based on data instead of guesses.

👉 In short: **Statistics = Data → Information → Decisions**

---

## 2. Types of Statistics

### a) Descriptive Statistics
- Summarizes and describes data.
- Answers: *“What does the data show?”*
- Examples:
  - Average marks of students in a class.
  - Percentage of people who prefer tea vs coffee.
- Tools: **Mean, Median, Mode, Range, Charts, Tables**

### b) Inferential Statistics
- Uses a sample of data to make conclusions about a larger group.
- Answers: *“What can we predict or conclude from the data?”*
- Examples:
  - Predicting election results based on exit polls.
  - Testing whether a new medicine is effective.
- Tools: **Hypothesis Testing, Confidence Intervals, Regression**

---

## 3. Real-Life Examples

1. **Classroom Example**
   - Collecting marks of 30 students in a test.
   - Descriptive: Average = 75 marks.
   - Inferential: Predicting that future students will also score around 75.

2. **Sports Example**
   - Descriptive: Virat Kohli’s batting average = 57.
   - Inferential: Predicting his probability of scoring a half-century in the next match.

3. **Health Example**
   - Survey: 100 people, 70 prefer jogging, 30 prefer yoga.
   - Descriptive: 70% prefer jogging.
   - Inferential: Conclude that in the whole city, majority may prefer jogging.

#### Mean and Median of Data

```python
import pandas as pd

# Create a sample DataFrame
data = {'col1': [10, 20, 30, 40, 50],
        'col2': [5, 15, 25, 35, 45]}
df = pd.DataFrame(data)

# Calculate the mean of each numerical column
mean_values = df.mean()
print("Mean values per column:\n", mean_values)

# Calculate the median of each numerical column
median_values = df.median()
print("\nMedian values per column:\n", median_values)
```

#### Mean and Median of Specific Column

```python
import pandas as pd

# Create a sample DataFrame
data = {'col1': [10, 20, 30, 40, 50],
        'col2': [5, 15, 25, 35, 45]}
df = pd.DataFrame(data)

# Calculate the mean of 'col1'
mean_col1 = df['col1'].mean()
print("Mean of 'col1':", mean_col1)

# Calculate the median of 'col2'
median_col2 = df['col2'].median()
print("Median of 'col2':", median_col2)
```
