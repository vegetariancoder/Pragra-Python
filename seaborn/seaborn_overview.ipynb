# 📊 Statistics Lecture: Understanding Variables

## 🎯 Overview



In this session, we discussed the concept of **variables** in statistics and explored the different **types of variables** with simple definitions and real-life examples.

---

## 🔍 What is a Variable?

A **variable** is a **property that can take up any value**.

> Example:
> - Age = 25
> - Gender = Male
> - Height = 5.9 ft

A variable holds **one value at a time**, and **a list of values is not a single variable**.

---

## 🧠 Types of Variables

Variables are mainly classified into:

### 1. Quantitative Variables
Variables that represent **numerical values**.

#### 🔹 a. Discrete Variables
- Can take **only whole numbers** (no fractions or decimals).
- Examples:
  - Number of children
  - Number of bank accounts
  - Number of students in a class
  - Number of workers in a company

#### 🔹 b. Continuous Variables
- Can take **any value**, including **fractions/decimals**.
- Examples:
  - Height (e.g., 175.5937363 cm)
  - Weight (e.g., 72.3 kg)
  - Age (e.g., 25.5 years)

---

### 2. Qualitative (Categorical) Variables
Variables that describe **categories** or **labels**, not numerical values.

- Examples:
  - Gender: Male, Female
  - Colors: Red, Green, Blue
  - States or Locations
  - Boolean: True, False

---

| Qualitative / Categorical | -        | Describes categories, no numeric operations      | Male/Female, Red/Blue, States     |

## Seaborn Plot types

- Seaborn provides a wide range of plot types that can be used for data visualization and exploratory data analysis. Broadly speaking, any visualization can fall into one of the three categories.

1. Univariate 🔹 – x only (This means we are looking at only one variable. Example: distribution of ages, salaries, exam scores.)
2. Bivariate 🔹 – x and y (Relationship between two variables. Example: height vs weight, hours studied vs marks scored.)
3. Trivariate 🔹 – x, y, z (When we bring in a third variable. Usually shown with color, size, or shape.)

Here are some of the most commonly used plot types in Seaborn:

1. **Scatter Plot**. A scatter plot is used to visualize the relationship between two variables. Seaborn's scatterplot() function provides a simple way to create scatter plots.
2. **Line Plot**. A line plot is used to visualize the trend of a variable over time. Seaborn's lineplot() function provides a simple way to create line plots.
3. **Histogram**. A histogram is used to visualize the distribution of a variable. Seaborn's histplot() function provides a simple way to create histograms.
4. **Box Plot**. A box plot is used to visualize the distribution of a variable. Seaborn's boxplot() function provides a simple way to create box plots.
5. **Violin Plot**. A violin plot is similar to a box plot, but provides a more detailed view of the distribution of the data. Seaborn's violinplot() function provides a simple way to create violin plots.
6. **Heatmap**. A heatmap is used to visualize the correlation between different variables. Seaborn's heatmap() function provides a simple way to create heatmaps.
7. **Pairplot**. A pairplot is used to visualize the relationship between multiple variables. Seaborn's pairplot() function provides a simple way to create pairplots.

### Aesthetic Control:

- Themes: darkgrid, whitegrid, dark, white, ticks.
- Palettes: Qualitative (e.g., deep, pastel), sequential (e.g., Blues), or diverging (e.g., coolwarm).

**In Seaborn:**
You can create histograms using the `sns.histplot()` function.

## 🔹 When to Use Histograms

- Use when you want to **understand the distribution** of a numeric variable.
- **Examples:** Exam scores, salaries, ages, or transaction amounts.
- Useful to detect patterns such as:
  - Normal distribution (bell curve)

## 🔹 Bins in Histograms

**Definition:**
Bins are the **intervals or ranges** into which the data is divided in a histogram.
Each bin represents a range of values, and the height of the bar shows how many data points fall inside that range.

---

**Example:**
Suppose you have exam scores:

`[45, 50, 52, 58, 65, 68, 72, 80, 85, 90]`

- If `bins=5`, the range of scores will be split into **5 equal intervals**
  *(like 40–50, 51–60, etc.)*
- If `bins=10`, you get **narrower intervals**
  *(like 40–45, 46–50, etc.)*

```python
# customize the histogram
import matplotlib.pyplot as plt
sns.set_style("whitegrid")
sns.histplot(data=iris_df, x="petal_length", bins=80 ,color="blue")
# add labels and title

plt.xlabel("Petal Length (in cm)")

plt.ylabel("Frequency (counts) ")

plt.title("Distribution of Petal Lengths in Iris Flowers")
```

## Seaborn Density Plots

#### Definition
Density plots, also known as **Kernel Density Plots (KDE plots)**, are a type of data visualization that display the **distribution of a continuous variable**.
Unlike histograms (which group data into bins), density plots provide a **smooth continuous curve** that estimates the probability distribution of the variable.

---

### 🔹 When to Use Density Plots
- To **visualize the shape of the data distribution** (normal, skewed, multimodal).
- To **compare multiple distributions** by plotting them together.
- When you want a **smoothed representation** of data instead of raw frequency counts.
- Useful in **exploratory data analysis (EDA)** to quickly see patterns and overlaps.

---

### 🔹 Why Choose Density Plots
- **Smoother than histograms:** Helps detect distribution shape more clearly.
- **Good for comparison:** Overlaying multiple density plots shows differences between groups.
- **Less sensitive to bin size:** Unlike histograms, KDE doesn’t depend on the number of bins, but on a parameter called **bandwidth** (controls smoothness).

---

### 🔹 Key Concepts
- **Bandwidth:** Determines how smooth the curve will be. Smaller bandwidth = more wiggly, larger bandwidth = smoother.
- **Area under curve:** Represents probability and always sums to 1.
- **Overlay on histogram:** Often combined with histograms for both raw counts and smooth distribution view.

## Seaborn Box Plots

#### Definition
Box plots (also known as **whisker plots**) are a type of visualization that show the **distribution of a dataset**.
They summarize key statistics about the data, including **median, quartiles, and outliers**.

---

### 🔹 When to Use Box Plots
- To understand the **spread and variability** of a numeric variable.
- To quickly **detect outliers** (values that lie far outside the main distribution).
- To **compare distributions across categories**, e.g., salaries by department, test scores by gender.
- To get a **five-number summary** (minimum, Q1, median, Q3, maximum).

---

### 🔹 Why Choose Box Plots
- **Compact summary:** Provides a lot of statistical information in a small chart.
- **Comparison-friendly:** Great for comparing distributions between groups side by side.
- **Outlier detection:** Highlights unusual data points clearly.
- Compared to **histograms** or **density plots**, box plots are less about the shape and more about **summary + spread**.

---

### 🔹 Key Concepts in a Box Plot
- **Box:** Represents the interquartile range (IQR = Q3 − Q1).
- **Line inside the box:** The **median** (50th percentile).
- **Whiskers:** Extend to the smallest and largest values within 1.5×IQR.
- **Outliers:** Points outside the whiskers are plotted individually.

## 🔹 Seaborn Violin Plots

#### Definition
A **violin plot** is a type of data visualization that combines aspects of both **box plots** and **density plots**.
It displays:
- A **density estimate** of the data (smoothed using a kernel density estimator).
- The **median** and **interquartile range (IQR)** in a box plot-like form inside the violin.

---

### 🔹 When to Use Violin Plots
- To visualize the **distribution and spread** of data along with **summary statistics**.
- To **compare multiple categories** side by side (e.g., exam scores by gender).
- When you want to see **both shape (density) and summary (quartiles, median)** in one plot.
- Useful in **exploratory data analysis (EDA)** to detect skewness, multimodal distributions (data with multiple peaks), and outliers.

---

### 🔹 Why Choose Violin Plots
- **More informative than box plots alone**: Shows both statistical summary and full distribution shape.
- **Better than histograms/density plots** when comparing multiple categories compactly.
- Useful for **highlighting multimodal distributions** (multiple peaks) that box plots may miss.
- Helps identify **asymmetry and skewness** visually.

---

### 🔹 Key Concepts in a Violin Plot
- **Wider sections** of the violin = more data points (higher density).
- **Narrow sections** = fewer data points.
- **Box plot inside the violin** = shows median and IQR.
- **Symmetry**: The plot is mirrored on both sides for easier interpretation.

## Seaborn Heatmaps

#### Definition
A **heatmap** is a graphical representation of data that uses **colors** to depict the values of a variable in a **two-dimensional space**.
The intensity of the color corresponds to the magnitude of the value.

---

### 🔹 When to Use Heatmaps
- To **visualize correlations** between variables in a dataset.
- To **spot patterns, trends, or anomalies** across rows and columns.
- To make large amounts of data easier to interpret quickly.
- To compare how different variables interact with each other (e.g., correlation matrices).

---

### 🔹 Why Choose Heatmaps
- **Color intensity makes interpretation intuitive** — easy to spot high/low values.
- More effective than raw tables of numbers for identifying relationships.
- Helpful in **exploratory data analysis (EDA)** when looking at relationships between numeric variables.
- Can highlight **clusters** or **groupings** of variables with strong correlation.

---

### 🔹 Key Concepts in Heatmaps
- **Correlation Matrix:** Often used with heatmaps, showing pairwise correlation (values between -1 and +1).
  - `+1` = perfect positive correlation
  - `-1` = perfect negative correlation
  - `0` = no correlation
- **Color Map (cmap):** Defines the color gradient (e.g., `coolwarm`, `viridis`).
- **Annotations:** Display numeric values inside the heatmap cells for clarity.

### Best practices for Seaborn visualization

1. Choose the right plot type for your data

    - Seaborn provides a wide range of plot types, each designed for different types of data and analysis.
    - It's important to choose the right plot type for your data to effectively communicate your findings.
    - For example, a scatter plot may be more appropriate for visualizing the relationship between two continuous variables, while a bar plot may be more appropriate for visualizing categorical data.

2. Use color effectively

    - Color can be a powerful tool for data visualization, but it's important to use it effectively.
    - Avoid using too many colors or overly bright colors, as this can make the visualization difficult to read. Instead, use color to highlight important information or to group similar data points.

3. Label your axes and use clear titles

    - Labels and titles are essential for effective data visualization.
    - Make sure to label your axes clearly and provide a descriptive title for your visualization. This will help your audience understand the message you are trying to convey.

4. Consider the audience

    - When creating visualizations, it's important to consider the audience and the message you are trying to communicate.
    - If your audience is non-technical, use clear and concise language, avoid technical jargon, and provide clear explanations of any statistical concepts.

5. Use appropriate statistical analysis

    - Seaborn provides a range of statistical functions that you can use to analyze your data.
    - When choosing a statistical function, make sure to choose the one that is most appropriate for your data and research question.
