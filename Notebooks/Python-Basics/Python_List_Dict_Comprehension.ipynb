# Python List and Dictionary Comprehension

Welcome! In this notebook, we will learn about two powerful and concise ways to create collections in Python: **List Comprehension** and **Dictionary Comprehension**.

## 1. List Comprehension

List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list (or any iterable).

### Why use it?
- It makes code more readable and concise.
- It is often faster than using a traditional `for` loop with `.append()`.

### Syntax:
 ```python
new_list = [expression for item in iterable if condition == True]
 ```

### Example 1: Creating a list of squares

````python
# Traditional way using for loop
numbers = [1, 2, 3, 4, 5]
squares = []
for n in numbers:
    if n % 2 == 0:
        squares.append(n**2)

print("Using for loop:", squares)
````

````python
square = [n**2 for n in numbers if n % 2 == 0]
squares
````

### Example 2: List comprehension with a condition (Filtering)
Let's create a list of even numbers from 0 to 9.

````python
# Get even numbers from range(10)
evens = [x for x in range(10) if x % 2 != 0]
print("Even numbers:", evens)
````

## 2. Dictionary Comprehension

Similar to list comprehension, dictionary comprehension allows us to create dictionaries in a single line of code.

### Syntax:
 ```python
new_dict = {key_expression: value_expression for item in iterable if condition == True}
 `` `

### Example 1: Creating a dictionary of squares
Let's map numbers to their squares.

````python
numbers = [1, 2, 3, 4, 5, 5, 4,2]
new_dict = {}

for n in numbers:
    new_dict[n] = numbers.count(n)

new_dict
````

````python
numbers = [1, 2, 3, 4, 5, 5, 4,2]
square_dict = {n : numbers.count(n) for n in numbers}

print("Square Dictionary:", square_dict)
````

### Example 2: Dictionary comprehension with a condition
Let's create a dictionary of only even numbers and their squares.

````python
numbers = [1, 2, 3, 4, 5, 5, 4,2]
even_square_dict = {x: x**2 for x in numbers if x % 2 == 0}
print("Even Squares Dictionary:", even_square_dict)
````
