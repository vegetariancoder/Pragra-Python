### Python Inheritance
* Inheritance is a way of creating a new class for using details of an existing class without modifying it.

* The newly formed class is a derived class (or child class). Similarly, the existing class is a base class (or parent class).

#### Types of Inheritance:

- Single Inheritance: A child class inherits from a single parent class.
- Multiple Inheritance: A child class inherits from more than one parent class.
- Multilevel Inheritance: A child class inherits from a parent class, which in turn inherits from another class.
- Hierarchical Inheritance: Multiple child classes inherit from a single parent class.

2. Multilevel Inheritance : A child class inherits from a parent class, which itself inherits from another parent class. It forms a chain (Grandparent → Parent → Child).

```python
class LivingThing:
    def breathe(self):
        return "Breathing..."

class Animal(LivingThing):
    def eat(self):
        return "Eating..."

class Dog(Animal):
    def bark(self):
        return "Barking..."

# Testing
my_dog = Dog()
print(my_dog.breathe())
```

3. Hierarchical Inheritance : In hierarchical inheritance, multiple child classes inherit from a single parent class.

```python
class Animal:
    def eat(self):
        return "Eating..."

class Dog(Animal):
    def bark(self):
        return "Woof!"

class Cat(Animal):
    def meow(self):
        return "Meow!"

# Testing
dog = Dog()
cat = Cat()
print(dog.eat(), dog.bark())
print(cat.eat(), cat.meow())
```

4. Multiple Inheritance - A single child class inherits from more than one parent class.

```python
class Flyer:
    def fly(self):
        return "Flying high!"

class Swimmer:
    def swim(self):
        return "Swimming fast!"

# Child inherits from both Flyer and Swimmer
class Duck(Flyer, Swimmer):
    def quack(self):
        return "Quack!"

# Testing
donald = Duck()
print(donald.fly())   # From Flyer
print(donald.swim())  # From Swimmer
print(donald.quack()) # From Duck
```

### Polymorphism

Polymorphism means "many forms". It refers to the ability of an entity (like a function or object) to perform different actions based on the context.

Technically, in Python, polymorphism allows same method, function or operator to behave differently depending on object it is working with. This makes code more flexible and reusable.

#### Real Life Example: In a backend payment system, multiple payment options are available such as Credit Card, UPI, NetBanking and Wallet. All payment types use a common method named processPayment() but different implementations:

- Credit Card Payment: validates card, talks to bank API
- UPI Payment: redirects to UPI gateway
- Wallet Payment: checks wallet balance
- NetBanking Payment: redirects to bank login

```python
class Animal:

    def sound(self):
        return "Some generic sound"

class Dog(Animal):

    def sound(self):
        return "Bark"

class Cat(Animal):

    def sound(self):
        return "Meow"


cat_obj = Cat()
print(cat_obj.sound())

dog_obj = Dog()
print(dog_obj.sound())

```

#### The super() function

In Python acts as a bridge between a child class and its parent class. It allows you to call methods and access attributes from the parent class without hardcoding the parent's name.


The two most common reasons to use super() are:
1. Initializing parent attributes inside a child's constructor (__init__).
2. Extending or overriding parent methods without rewriting the original code.

1. Using super() in __init__ (Constructor Inheritance)


When a child class creates its own __init__ method, it overrides the parent's __init__. Without super(), the parent's attributes won't be initialized. super() lets you pass data up to the parent class.

```python
class Person:
    def __init__(self, name, age):
        print("--> Inside Person.__init__")
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, student_id):
        print("--> Starting Student.__init__")
        # Call the parent's __init__ method using super()
        super().__init__(name, age)
        print("--> Returned back to Student.__init__")
        self.student_id = student_id

# Testing
student1 = Student("Sahil", 30, "S12345")
print(student1.name)       # Inherited from Person
print(student1.age)        # Inherited from Person
print(student1.student_id) # Defined in Student
```

2. Using super() to Extend Methods

Sometimes you want to modify a parent method, but you still want the parent's core behavior to run. super() lets you execute the parent's method and add extra code on top of it.

```python
class Employee:
    def get_details(self):
        return "Employee Details: "

class Manager(Employee):
    def get_details(self):
        # Get the parent's string first using super()
        base_details = super().get_details()
        # Extend it with manager-specific details
        return base_details + "Role: Manager"

# Testing
mgr = Manager()
print(mgr.get_details())
# Output: Employee Details: Role: Manager
```

### What is Encapsulation in Python?

- Encapsulation is a technique for hiding a class’ internal operations and exposing only the information required for external interactions.
- It enables the creation of objects with clearly defined behaviors and properties, enhancing their usability and security.

#### Reasons to use encapsulation in python:
1. Only the necessary information is exposed and the implementation details are concealed from the outside world, the code is made more secure.
2. It allows for the creation of objects with clearly defined properties and behaviors, which facilitates the reuse of the code throughout the program.

### Private access

Private variables and methods are those that can only be accessed within the class. In Python, private variables and methods are declared by prefixing them with double underscores (`__`).

```python
class Car:

    __private_capacity = "2000cc"
    public_capacity = "1000cc"

    def __start_engine(self):
        print("Engine started!")

    def drive(self):
        print("Engine driving!")
        return self.__private_capacity


car_obj_1 = Car()

```

#### Name Mangling

- Python provides a magic wand that can be used to call private methods outside the class also, it is known as name mangling.
- It means that any identifier of the form __method (at least two leading underscores or at most one trailing underscore) is replaced with _classname__method, where the class name is the current class name with a leading underscore(s) stripped.

```python
class Car:

    __private_capacity = "2000cc"
    public_capacity = "1000cc"

    def __start_engine(self):
        print("Engine started!")

    def drive(self):
        print("Engine driving!")

carobj = Car()

# name mangling

carobj._Car__start_engine()
```
