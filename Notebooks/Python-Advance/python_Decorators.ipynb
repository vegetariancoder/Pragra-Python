### Python Decorators

- In Python, a decorator is a `design pattern` that allows you to modify the functionality of a function by wrapping it in another function.
- Decorators are a powerful and flexible way to modify or extend the behavior of functions or methods, without changing their actual code.
- A decorator is essentially a function that takes another function as an argument and returns a new function with enhanced functionality.




### Prerequisites for learning decorators
- Before we learn about decorators, we need to understand a few important concepts related to Python functions.
- Also, remember that everything in Python is an object, even functions are objects.

### Nested Function

```python
def addition(number1,number2):
    return number1 + number2

myVar = addition(1,2)
myVar
```

```python
def Outer(number1):
    def Inner(number2):
        return number1 + number2
    return Inner



firstFunction = Outer(number1=3)
print(firstFunction)
#
additionResult = firstFunction(number2=5)
print(additionResult)
```

### Pass Function as Argument

```python
def calculate(func_name, number1, number2,number3=0):
    return func_name(number1, number2,number3)
```

```python
def add(x, y):
    return x + y

def simpleInterest(p, r, t):
    return (p * r * t)/100

def calculate(func_name,number1, number2,number3=0):
    if func_name == add:
        return func_name(number1, number2)
    elif func_name == simpleInterest:
        return func_name(number1, number2,number3)
    return None


adding = calculate(func_name = add, number1 = 4, number2 = 6)
print(adding)


si = calculate(func_name = simpleInterest, number1 = 4, number2 = 6, number3 = 8)
print(si)
```

### Now comes the most irritating yet most asked thing : DECORATORS !!! 😂


- Decorators let you modify or extend function behavior without changing the original code
- Use the @decorator_name syntax to apply decorators cleanly

```python
def make_pretty(func):
    def inner():
        print("I got decorated")
        func()
    return inner


def ordinary():
    print("I am ordinary")

decorated = make_pretty(ordinary)
print(decorated)
```

### Using Decorators

```python
def make_pretty(func):
    def inner():
        func()
        print("I got decorated")
        print("I am decorating all functions")

    return inner


@make_pretty
def ordinary():
    print("I am ordinary")


@make_pretty
def new_func():
    print("I am new function")

new_func()

```

### Some other examples of Decorators

### @staticmethod


The @staticmethod decorator is used to define a method that doesn’t operate on an instance of the class (i.e., it doesn’t use self). Static methods are called on the class itself, not on an instance of the class.

```python
class Operation:

    def addition(self,x,y):
        return x+y

    def subtraction(self,x,y):
        return x-y
```

```python
# instance method --> self
# class method --> cls
# static method --> NO IMPLICIT ARGUEMENT
```

```python
obj1 = Operation()
obj1.addition(x=7,y=8)
```

```python
class MathOperations:

    @staticmethod
    def add(x, y):
        return x + y

print(MathOperations.add(1, 2))
```

#### @staticmethod: Use when you need a utility function that performs a task related to the class but doesn’t need to access class or instance data.

### @classmethod

The @classmethod decorator is used to define a method that operates on the class itself (i.e., it uses cls). Class methods can access and modify class state that applies across all instances of the class.

```python
class Book:

    def __init__(self,name):
        self.name = name

    def print_book(self):
        print(self.name)

b1 = Book("Python 101")
b1
```

```python
# factory methods

class Book:
    total_books = 0

    def __init__(self, title):
        self.title = title # object
        Book.total_books += 1 # class

    @classmethod
    def get_total_books(cls):
        print(Book.total_books)
        return cls.total_books

class Book2(Book):
    pass

# Usage
book1 = Book("Python 101") # total_books = 1
book2 = Book("Data Science with Python") # total_books = 2


print(Book.get_total_books())  # Output: 2
```

#### Use @classmethod to keep track of class-level data

```python
class MyClass:
    category = "Educational"

    @classmethod
    def show_category(cls):
        return f"This class category is: {cls.category}"

# 1. Accessing via Class (Standard way)
print(MyClass.show_category())

# 2. Accessing via Object/Instance (Works perfectly)
my_object = MyClass()
print(my_object.show_category())
```

#### Key Behaviors to Note

1. No Access to self: Even though you call the method using an object (my_object.show_category()), the method still only receives the class (cls). It cannot access any instance variables or attributes tied to self.
2. Best Practice: While Python allows accessing class methods via an object, it is generally considered cleaner and more readable to call them directly using the class name (ClassName.method_name()) to avoid confusing them with regular instance methods.
