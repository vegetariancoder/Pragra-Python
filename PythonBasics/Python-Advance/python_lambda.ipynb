#### Python Lambda Functions

#### Lambda functions in Python are powerful, concise tools for creating small, anonymous functions on the fly.

- Lambda functions differ from standard Python functions in several key ways. They are anonymous expressions, meaning they have no name unless explicitly assigned to a variable.
- They are also more concise and defined in a single line without the need for a return statement.

```python
sq_lambda = lambda x : x**2
```

```python
cub_lambda = lambda x : x**3
```

```python
print(cub_lambda(4))
```

#### Use Cases
1. Condition Checking: lambda function can use conditional expressions (if-else) to return different results based on a condition.

```python
check = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
```

2. Returning Multiple Results: Although a lambda can contain only one expression, it can still return multiple results by combining them into a tuple.

```python
calc = lambda x, y: (x + y, x * y)
res = calc(3, 4)
print(res)
```

#### Common use cases for lambda functions

Lambda functions are often used in functional programming, particularly with functions like `map()` and `filter()`,

Filter : filter() function is used to extract elements from an iterable (like a list, tuple or set) that satisfy a given condition. It works by applying a function to each element and keeping only those for which function returns True.

```python
def is_even(number):
    return number % 2 == 0


numbers = [1, 2, 3, 4, 5, 6, 7, 8]
result = filter(is_even, numbers)
print(tuple(result))
```

```python
for nums in numbers:
    is_even(nums)
```

```python
# Use filter with lambda function
numbers = [1, 2, 3, 4, 5, 6, 7, 8]


evens = filter(lambda x: x % 2 == 0, numbers)


print(list(evens))
```

Map : map() function in Python applies a given function to each element of an iterable (list, tuple, set, etc.) and returns a map object (iterator). It is a higher-order function used for uniform element-wise transformations, enabling concise and efficient code.

```python
s = ['1', '2', '3', '4']
new_s = []

for element in s:
    new_s.append(int(element))
```

```python
s = ['1', '2', '3', '4']
res = map(int, s)
print(list(res))
```

```python
# Use map with lambda function
fruits = ['apple', 'banana', 'cherry']
lengths = map(lambda x: len(x), fruits)
print(list(lengths))
```

#### Q: The Salary Increment Tool

##### Scenario:
The HR department wants to give a quick $500 bonus to several employees. Instead of writing a full def function for this simple math operation, you want to use a Lambda function to handle the calculation instantly.


##### Your Task:
- Create a lambda function called add_bonus.
- The function should take one input (salary) and return that salary plus 500.
- Test your lambda function by passing in the salary of person (10000).

```python
def add_bonus(salary):
    return salary + 500

add_bonus(10000)
```

```python
add_bonus = lambda salary : salary + 500

add_bonus(10000)
```
