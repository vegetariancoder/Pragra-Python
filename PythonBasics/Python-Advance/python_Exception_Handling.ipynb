# Python Exceptions

### An exception is an unexpected event that occurs during program execution. For example,

`divide_by_zero = 7 / 0`

### The above code causes an exception as it is not possible to divide a number by 0.

### Python Logical Errors (Exceptions)
#### Errors that occur at runtime (after passing the syntax test) are called exceptions or logical errors.

### For instance, they occur when we

* try to open a file(for reading) that does not exist (FileNotFoundError)
* try to divide a number by zero (ZeroDivisionError)
* try to import a module that does not exist (ImportError) and so on.

Whenever these types of runtime errors occur, Python creates an exception object.

If not handled properly, it prints a traceback to that error along with some details about why that error occurred.

### Multiple except statement example

- In this example, we will be using multiple except statements to handle multiple types of exceptions.
- If a ZeroDivisionError exception is raised, the program will print "You cannot divide a value with zero."
- The rest of the exceptions will print “Something else went wrong.”

```python
try:
   print(1/0)
except ZeroDivisionError:
   print("You cannot divide a value with zero")
except:
   print("Something else went wrong")
```

### Python try with else clause

<img src="../../images/tryexceptelsePython.png" alt="try except diagram" width="500" height="100" />

- When the try statement does not raise an exception, code enters into the else block.
- It is the remedy or a fallback option when you expect a part of your script will produce an exception.
- It is generally used in a brief setup or verification section where you don't want certain errors to hide.

```python
try:
   result = 1/34
   print(f"Your answer is {result}")
except ZeroDivisionError:
   print("Our Error Message")
else:
   print(f"Your answer is {result}")
```

##### IndexError with else example

```python
x = [5,8,9,13]

def find_nth_value(mylist,n_index):
    try:
        result = mylist[n_index]
    except IndexError as err:
        print(err)
    else:
        print("Your answer is ", result)
```

```python
find_nth_value(mylist=x,n_index=2)
```

### The finally keyword in Python

- The finally keyword in the try-except block is always executed, irrespective of whether there is an exception or not.
- In simple words, the finally block of code is run after the try, except and else block is final.
- It is quite useful in cleaning up resources and closing the object, especially closing the files.

<img src="../../images/tryexceptelsefinallyinPython.png" alt="try except diagram" width="500" height="90" />

```python
def divide(x,y):
    try:
        result = x/y
    except ZeroDivisionError:
        print("Please change 'y' argument to non-zero value")
    else:
        print(f"Your answer is {result}")
    finally:
        print("<---This is always going to execute--->")
```

```python
def divide(x,y):
    try:
        result = x/y
    except ZeroDivisionError:
        print("Please change 'y' argument to non-zero value")
    finally:
        print("<---This is always going to execute--->")
```

```python
divide(x=1,y=2)
```

#### Python `Raise` Keyword - Raising Exceptions in Python

- As a Python developer, you can throw an exception if certain conditions are met. It allows you to interrupt the program based on your requirements.
- The `raise` keyword in Python is used to manually trigger an exception.
- It tells Python: “Hey, throw this error right now!”
- You can raise built-in exceptions (ValueError, ZeroDivisionError, etc.) or your own custom exceptions.

```python
age = -23
if age < 0:
    raise ValueError("Age cannot be negative")
else:
    print("Age is valid")
print("Program continues")

# Send a notification email --> data
```

```python
value = 1001
try:
    if value > 1000:

        # raise the ValueError
        raise ValueError("Please add a value lower than 1,000")
    else:
        print("Congratulations! You are the winner!!")

except ValueError as e:
        print(e)
```

### Summary:
✅ Use `raise` when you want to create/throw an error.

✅ Use `except` when you want to catch/handle an error gracefully.

✅ Together, they let you control error flow instead of crashing.


#### Think of it like this:
`raise` = "I'm reporting a problem!"

`except` = "I'll handle that problem"

### Defining Custom Exceptions
In Python, we can define custom exceptions by creating a new class that is derived from the built-in `Exception` class.

```python
def test():
    pass
```

```python
class MyError(Exception):
    pass

try:
    raise MyError("This is a custom error")
except MyError as e:
    print(e)
```

```python
myList = [1, 2, 4, 5, 4, 6] # do not want 5 in my list

# logging
# customization for errors

class IHate5InMyList(Exception):
    pass

try:
    for element in myList:
        if element == 5:
            raise IHate5InMyList("Element cannot be 5 or I do not expect 5 in my list")
except IHate5InMyList as e:
    print(e)
```

#### Title: Simple Calculator with Custom Error Handling
Write a program that:
- Asks the user to input two numbers and an operation (`+`, `-`, `*`, `/`).
- Performs the calculation and displays the result.
##### Requirements:
- Use `try`, `except`, `else`, and `finally`.
- Handle:
    - Invalid number input
    - Division by zero
    - Raise a custom exception if the user enters an invalid operation (anything other than +, -, *, /).
- Use `else` to display the result if no exception occurs.
- Use `finally` to print "Thank you for using the calculator."

```python
class WrongOperatorError(Exception):
    pass

def calculate():
    try:
        num1 = float(input("Enter the first number: "))
        num2 = float(input("Enter the second number: "))
        operation = input("Enter an operation (+, -, *, /): ")


        if operation not in ['+', '-', '*', '/']:
            raise WrongOperatorError(f"'{operation}' is not a valid operation.")

        if operation == '+':
            result = num1 + num2
        elif operation == '-':
            result = num1 - num2
        elif operation == '*':
            result = num1 * num2
        elif operation == '/':
            result = num1 / num2

    except ValueError:
        print("Error: Invalid number input. Please enter numerical values.")

    except ZeroDivisionError:
        print("Error: Division by zero is not allowed.")

    except WrongOperatorError as e:
        print(f"Error: {e}")

    else:
        print(f"The result of {num1} {operation} {num2} is: {result}")

    finally:
        # Executes no matter what
        print("Thank you for using the calculator.")
```
