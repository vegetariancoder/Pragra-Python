### Types of Python Function Arguments

##### Python supports various types of arguments that can be passed at the time of the function call. In Python, we have the following function argument types in Python:

1. Default argument
2. Keyword arguments (named arguments)
3. Positional arguments
4. Arbitrary arguments (variable-length arguments *args and **kwargs)

#### Default Arguments

* A default argument is a parameter that assumes a default value if a value is not provided in the function call for that argument. The following example illustrates Default arguments to write functions in Python.

````python
def myFridayNight(food="Pizza", movie='Harry Potter', drink = "Coke"):
    print("I will eat : ", food)
    print("I will watch : ", movie)
    print("I will drink : ", drink)


print(myFridayNight(movie="Family Man",drink="Pepsi"))
# myFridayNight(food="Pizza")
````

#### Keyword Arguments

* The idea is to allow the caller to specify the argument name with values so that the caller does not need to remember the order of parameters.

````python
def myInfo(name, age,city,color = "black",movie="Harry Potter"):
    myName = "my name is ", name
    myAge = "my age is ", age
    myCity = "my city is ", city
    myColor = "my fav color is ", color
    return myName, myAge, myCity, myColor
````

````python
myInfo(name="sahil",age=29)
````

````python
myInfo(age=28, city="Delhi",name="Sahil")
````

````python
def student(firstname, lastname, middlename="James"):
    print("The first name is:",firstname, ",the middle name is:",middlename, "and the last name is:",lastname)

student(middlename="Snape", firstname="Harry", lastname="Potter")
````

#### Arbitrary Keyword Arguments

* It refers to the ability of a function to accept an unspecified number of keyword arguments. This is achieved using the `*args` and `**kwargs` syntax within the function's parameter list.

#### Why do we need Arbitrary Keyword Arguments?

* Sometimes, you might not know in advance how many keyword arguments a function will receive.
* `*args` and `**kwargs` allows you to collect these arguments into a list or dictionary within the function.
* The `*args` variable becomes a list,string,tuple etc and the `*kwargs` variable becomes a dictionary where the keys are the argument names and the values are the argument values.

#### *args (Positional Arguments)

Use `*args` when you want to pass a variable number of non-keyworded arguments (a list of items). Inside the function, args becomes a tuple.

````python
nums = [1,2,3,4]

sum_val = 0

for elements in nums:
    sum_val += elements

print(sum_val)
````

````python
def addingElementsList(*args):
    total = 0
    for elements in args:
        total = total + elements
    return total
````

````python
addingElementsList(1,2,3,4,56,76,3,5,56.9,23,67,3,2,4,5,6,7)
````

````python
def sumList(myList):
    total = 0
    for numbers in myList:
        total += numbers
    return total


myList = [1, 2, 3, 4, 5,8,3,4,2,43]
print(type(myList))
sumList(myList=myList)
````

#### **kwargs (Keyword Arguments)

Use `**kwargs` when you want to pass a variable number of keyworded arguments (named variables like key="value"). Inside the function, kwargs becomes a dictionary.

````python
def student_info(**kwargs):
    print(kwargs)


student_info(name="John", age=20, grade="A", learning="Python", myKey=90)
````

````python
def student_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")


student_info(name="Bob", age=19, subject="Math", myKey=80,myfavstudent="Tarun")
````

````python
def student_info(**kwargs):
    print("Name:", kwargs.get("name"))
    print("Age:", kwargs.get("age"))


student_info(name="Alice", age=22)
````

Write a Python function called `process_data` that:

* Takes two required positional arguments, `operation` (a string like "add" or "multiply") and `initial_value` (a number).
* Uses `*args` to accept any number of additional numbers.
* Uses `**kwargs` to accept any number of optional keyword arguments, but you only need to process one specific keyword argument, "message".
* If operation is "add", add all the *args numbers to initial_value and return the result.
* If operation is "multiply", multiply all the *args numbers with initial_value and return the result.
* If the "message" keyword argument is present, print it before returning the calculated result.

````python
def process_data(operation, initial_value, *args, **kwargs):
    result = initial_value
    if operation == "addition":
        for items in args:
            result = result + items
    elif operation == "subtraction":
        for items in args:
            result = result - items
    elif operation == "multiplication":
        for items in args:
            result = result * items

    if "message" in kwargs:
        print(kwargs["message"], end=" ")

    return result


print(process_data("multiplication", 5, 6, 7, 8, 9, 10, message="Result :"))
````
