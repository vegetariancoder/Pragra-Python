### Python Decorators

- In Python, a decorator is a `design pattern` that allows you to modify the functionality of a function by wrapping it in another function.
- Decorators are a powerful and flexible way to modify or extend the behavior of functions or methods, without changing their actual code.
- A decorator is essentially a function that takes another function as an argument and returns a new function with enhanced functionality.

### Now comes the most irritating yet most asked thing : DECORATORS !!!

```python
# A simple decorator function
def walmartsDecoration(func):

    def wrapper():
        print("Before calling the function.")
        func()
        print("After calling the function.")
    return wrapper

# Applying the decorator to a function
# @ --> decorator function

@walmartsDecoration # --> decorator
def myName(): # --> function to decorate
    print("My name is Sahil.")


# print(greet())
myName()
```

```python
### decorator called -- logging(function)
### inside logging --> print("Sahil called function name")

### create a dedicated function -- printMyName() --> print("My name is Sahil.")
### Decorate printMyName with @logging
```

```python
def logging(func):
    def wrapper_function():
        print("Sahil called :", func.__name__)
        func()
    return wrapper_function

@logging
def printMyName():
    print("My name is Sahil.")

@logging
def printSunnyName():
    print("My name is Sunny.")

printMyName()
printSunnyName()
```

### Decorating Functions with Parameters

```python
def smart_divide(func):
    def inner(a, b):
        print("I am going to divide", a, "and", b)
        if b == 0:
            print("Whoops! cannot divide")
            return None

        return func(a, b)
    return inner

@smart_divide
def divide(a, b):
    print(a/b)


# print(divide(2, 5))

print(divide(2, 0))
```

### Some other examples of Decorators

### @staticmethod

The @staticmethod decorator is used to define a method that doesn’t operate on an instance of the class (i.e., it doesn’t use self). Static methods are called on the class itself, not on an instance of the class.

```python
class Operation:

    def addition(self,x,y):
        return x+y

    def subtraction(self,x,y):
        return x-y

op = Operation()
print(op)
op.addition(1,2)
```

```python
class MathOperations:
    myname = "sahil"

    @staticmethod
    def add(x, y):
        return x + y


# Using the static method

# obj=MathOperations()
# add_value = obj.add(1,2)
# print(add_value)

add_value = MathOperations.add(1, 2)
#
print(add_value)
```

### @classmethod

The @classmethod decorator is used to define a method that operates on the class itself (i.e., it uses cls). Class methods can access and modify class state that applies across all instances of the class.

```python
class Employee:
    raise_amount = 1.05 # global variable

    @classmethod
    def set_raise_amount(cls, amount):
        cls.raise_amount += amount

# Using the class method (without object)

print(Employee.raise_amount) # 1.05
Employee.set_raise_amount(amount=1.11)
print(Employee.raise_amount)
```

```python
import logging

# Set up basic logging (prints to console)
logging.basicConfig(level=logging.INFO)

# Simple logging decorator
def log_function(func):
    def wrapper(*args, **kwargs):
        logging.info(f"Running function: {func.__name__}")
        return func(*args, **kwargs)
    return wrapper


@log_function
def greet():
    print("Hello, bro!")

print(greet())
```
