### Python Package

- A package is a container that contains various functions to perform specific tasks. For example, the math package includes the `sqrt()` function to perform the square root of a number.
- While working on big projects, we have to deal with a large amount of code, and writing everything together in the same file will make our code look messy.
- Instead, we can separate our code into multiple files by keeping the related code together in packages.
- We can use the package whenever we need it in our projects. This way we can also reuse our code.

### Modular Calculator

- Create a Python calculator module with basic math operations and import it into other files to perform specific tasks.

<img src="../img/structure.png" alt="Assertion Example" width="400" height="300"/>

- Create `test_calculation.py` file and test all the functions using `assert`



```python
import pytest
from calculations import (
    add,
    subtract,
    multiply,
    divide,
    rectangle_area,
    simple_interest
)

# Addition tests
def test_add():
    assert add(2, 3) == 5
    assert add(-1, 1) == 0

# Subtraction tests
def test_subtract():
    assert subtract(10, 5) == 5
    assert subtract(0, 5) == -5

# Multiplication tests
def test_multiply():
    assert multiply(4, 5) == 20
    assert multiply(0, 100) == 0

# Division tests
def test_divide():
    assert divide(10, 2) == 5
    assert divide(9, 3) == 3

def test_divide_by_zero():
    with pytest.raises(ValueError, match="Cannot divide by zero"):
        divide(5, 0)

# Rectangle area test
def test_rectangle_area():
    assert rectangle_area(10, 2) == 20
    assert rectangle_area(5, 5) == 25

# Simple interest test
def test_simple_interest():
    assert simple_interest(1000, 5, 2) == 100
    assert simple_interest(2000, 3, 1) == 60
```
