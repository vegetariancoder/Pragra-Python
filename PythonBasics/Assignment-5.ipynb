```python
# Define a class Rectangle with methods to compute area and perimeter.

# granular - in pieces
# class Rectangle
# methods compute area and perimeter
```

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.height + self.width)

rect = Rectangle(5,8)
print(rect.area())
print(rect.perimeter())

```

```python
 # Write a class BankAccount with methods deposit() and withdraw(). Ensure balance never goes negative.
```

```python
class BankAccount:

    def __init__(self,balance=0):
        self.balance = balance

    def deposit(self,amount):
        """
        :param amount: integer amount to be deposited
        :return:
        """
        try:
            if type(amount) == 'int' and amount > 0:
                self.balance = self.balance + amount
        except ValueError:
            print("Amount is not integer")

    def withdraw(self,amount):
        """

        :param amount:
        :return:
        """
        if 0 < amount <= self.balance:
            self.balance = self.balance - amount
        else:
            print("Either the amount value is not int or insufficient amount")
```
