```python
# we check input - divide by every number between 2 and square root of that number
# if num == 0 ( 2 and sqrt of number) --> not prime

def check_prime(number):

    # prime number is greater than 1
    if number <=1 :
        return False
    for i in range(2,number):
        if number % i == 0:
            return False
    return True
```

### Read CSV Files with Python
- The csv module provides the csv.reader() function to read a CSV file.

```python
try:
    with open('/Users/sahilnagpal/Desktop/Pragra-Python/data/myFile.csv', mode='r') as myFile:
        reader = csv.reader(myFile)
        # print(type(reader))
        # print(reader)
        for rows in reader:
            print(rows)
except FileNotFoundError:
    print("File not found")
#
```

#### Frequently asked questions
1. How do I handle a large CSV file in Python?
- To handle large CSV files, use efficient techniques like reading the file line by line with Python’s built-in csv module or using the pandas library with chunking (pd.read_csv(..., chunksize=...)). These methods help manage memory usage and avoid performance issues.

2. Why is CSV used in Python?
- CSV files are easy to read and write, making them ideal for storing and exchanging structured data. Python supports CSV handling through built-in tools, allowing developers to work with tabular data quickly and efficiently.

3. How do I edit a CSV file in Python?
- To edit a CSV file, read its content into memory using the csv.reader() or pandas.read_csv(), modify the data as needed, and then write it back using csv.writer() or pandas.to_csv().

4. Is csv a standard Python library?
- Yes, the csv module is part of Python’s standard library. You can use it without any external installation by importing it with import csv.
