#### Functions – Predict the Output (5 Problems)

```python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Alice")
```

```python
def add(a, b):
    return a + b

def multiply(a, b):
    print(a * b)

x = add(2, 3)
y = multiply(2, 3)
print(x, y)
```

```python
x = 5

def change():
    x = 10
    print("Inside function:", x)


change()
print("Outside function:", x)
```

```python
def append_item(item, my_list=[]):
    my_list.append(item)
    return my_list # mylist = [1] # mylist = [1,2]

print(append_item(1))
print(append_item(2))
```

```python
class Animal:
    def speak(self):
        return "Sound"

class Dog(Animal):
    def speak(self):
        return "Bark"

obj1 = Animal()
obj2 = Dog()
print(obj1.speak(), obj2.speak())
```
