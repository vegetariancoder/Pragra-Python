```python
import mysql.connector
# install sql server

try:
  print("Connecting to MySQL database")  # logger messages connection
  with mysql.connector.connect(
          host="localhost",
          user="root",
          passwd="password"
  ) as dataBase:
    print("Connected to MySQL database")

    connection = dataBase.cursor()
    print("Creating database called : Pragra_Feb_Batch_Database")

    connection.execute("CREATE DATABASE IF NOT EXISTS Pragra_Feb_Batch_Database")

    print("Database created successfully")

    connection.execute("CREATE TABLE IF NOT EXISTS pragraNewDatabase.student (id INT AUTO_INCREMENT PRIMARY KEY, name VARCHAR(255), age INT)")

    print("Table created successfully")

    connection.execute("select * from pragraNewDatabase.student")

    myresult = connection.fetchall()

    for x in myresult:
      print(x)

except mysql.connector.Error as err:
  print(f"Error: {err}")



```
