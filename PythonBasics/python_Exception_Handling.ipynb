`divide_by_zero = 7 / 0`

### Python Logical Errors (Exceptions)
#### Errors that occur at runtime (after passing the syntax test) are called exceptions or logical errors.

### For instance, they occur when we

* try to open a file(for reading) that does not exist (FileNotFoundError)
* try to divide a number by zero (ZeroDivisionError)
* try to import a module that does not exist (ImportError) and so on.

Whenever these types of runtime errors occur, Python creates an exception object.

If not handled properly, it prints a traceback to that error along with some details about why that error occurred.

```python


try:
    divide_numbers = 90 / 0
    print(divide_numbers)
except ZeroDivisionError:
    print("You can not divide by zero")


# try --> # except


# try (try block) -> keyword where you will put all your logic
# except -> catch the exception and return a meaningful msg
```

### Catching Specific Exceptions in Python

#### For each `try` block, there can be zero or more `except` blocks. Multiple `except` blocks allow us to handle each `exception` differently.

#### The argument type of each except block indicates the type of exception that can be handled by it. For example,

```python
numbers = [2, 4, 6, 8]

print(numbers[4])
```

```python
numbers = [2, 4, 6, 8]
try:
    print(numbers[7])
except IndexError:
    print("Index out of range or not found")

# Output: Index Out of Bound
```

#### Python `Raise` Keyword

- The `raise` keyword in Python is used to manually trigger an exception.
- It tells Python: “Hey, throw this error right now!”
- You can raise built-in exceptions (ValueError, ZeroDivisionError, etc.) or your own custom exceptions.

```python
# define Python user-defined exceptions

class NotMinimumAgeException(Exception):
    pass  # this does nothing


number = 18

try:
    input_num = int(input("Enter a number: "))
    if input_num < number:
        raise NotMinimumAgeException
    else:
        print("Eligible to Vote")

except NotMinimumAgeException:
    print("Exception occurred: Invalid Age")

print(type(NotMinimumAgeException))
```
