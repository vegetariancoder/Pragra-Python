#### Function DocStrings

- In Python, the documentation string is also called a docstring. It is a descriptive text (like a comment) written by a programmer to let others know what block of code does.

##### Single-Line Docstring

The single-line docstring is a docstring that fits in one line. We can use the triple single or triple-double quotes to define it. The Opening and closing quotes need to be the same. By convention, we should use to use the triple-double quotes to define docstring.

````python
def addition(num1, num2):
    print(num1+ num2)

print(addition.__doc__)
````

````python

def factorial(x):
    """this is a factorial function doing blah blah"""
    return x

# access doc string
print(factorial.__doc__)
````

##### Multi-Line Docstring

A multi-line Docstrings is the same single-line Docstrings, but it is followed by a single blank line with the descriptive text.

````python
def any_fun(parameter1):
    """
    This is my finc
    :param parameter1: integer
    :return: blah blah
    """
    pass
print(any_fun.__doc__)
````

### Types of Python Function Arguments

##### Python supports various types of arguments that can be passed at the time of the function call. In Python, we have the following function argument types in Python:

1. Default argument
2. Keyword arguments (named arguments)
3. Positional arguments
4. Arbitrary arguments (variable-length arguments *args and **kwargs)

#### Default Arguments

* A default argument is a parameter that assumes a default value if a value is not provided in the function call for that argument. The following example illustrates Default arguments to write functions in Python.

````python
def myHoliday(food="Pizza",day="Saturday"):
    return f"I and eat {food} and day is {day}"

````

````python
myHoliday(food="Burger",day="Sunday")
````

````python
def myHoliday(movie, food="Pizza",day="Saturday"):
    return f"I am going to watch {movie} and eat {food} and day is {day}"
````

#### Arbitrary Keyword Arguments

* It refers to the ability of a function to accept an unspecified number of keyword arguments. This is achieved using the `*args` and `**kwargs` syntax within the function's parameter list.

#### Why do we need Arbitrary Keyword Arguments?

* Sometimes, you might not know in advance how many keyword arguments a function will receive.
* `*args` and `**kwargs` allows you to collect these arguments into a list or dictionary within the function.
* The `*args` variable becomes a list,string,tuple etc and the `*kwargs` variable becomes a dictionary where the keys are the argument names and the values are the argument values.

#### *args (Positional Arguments)

Use `*args` when you want to pass a variable number of non-keyworded arguments (a list of items). Inside the function, args becomes a tuple.

````python
nums = [1,2,3,4]

sum_val = 0

for elements in nums:
    print(elements)
    sum_val += elements

print(sum_val)
````

````python
def len_of_element(*args):
    sum_val = 0
    for elements in args:
        sum_val += elements

    return sum_val
````

````python
len_of_element(1,2,4,3,4,3,4,3,2,13,2,3,2)
````

````python
def sumList(myList):
    total = 0
    for numbers in myList:
        total += numbers
    return total


myList = [1, 2, 3, 4, 5,8,3,4,2,43]

sumList(myList=myList)
````

#### **kwargs (Keyword Arguments)

Use `**kwargs` when you want to pass a variable number of keyworded arguments (named variables like key="value"). Inside the function, kwargs becomes a dictionary.

````python
def student_info(**kwargs):
    print(kwargs)


student_info(name="John", age=20, grade="A", learning="Python", myKey=90,myfavstudent="Tarun")
````

````python
def student_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")


student_info(name="Bob", age=19, subject="Math", myKey=80,myfavstudent="Tarun", myfavsubject="Physics",firstname = "sahil")
````

````python
def student_info(**kwargs):
    print("Name:", kwargs.get("name"))
    print("Age:", kwargs.get("age"))


student_info(name="Alice", age=22)
````

````python
def student_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")
````
