### Python Class and Objects - Full Code

```python
class Dog:

    sound = "bark"  # class attribute

    def bark(self): # class method
        return f"Dogs make sound : {self.sound}"

    def eat(self,name):
        return "Dogs eat food"

# Creating object from class
dog1_object = Dog()

# variable
print(dog1_object.sound)

# functions
print(dog1_object.bark())
```

### Create Multiple Objects of Python Class

```python
class Dog:

    sound = "bark"  # class attribute

    def bark(self): # class method
        return f"Dogs make sound : {self.sound}"

# Creating object from class
dog1_object = Dog()
dog2_object = Dog()

# variable
print(dog1_object.sound)

# functions
print(dog2_object.sound)
```

### init method-constructor

- When you create an object from a class, it’s like building a house from a blueprint.
- The __init__() method is like the setup/initialization step where you decide the color of the walls, number of rooms, etc.
- The __init__() method is called automatically when you create an object from a class.
- Without __init__(), every house would be blank — you’d have to set values one by one after construction.

```python
# Simple Class Without __init__()

class Student:

    name = "Sahil"
    course = "Python"

    def learning(self):
        return f"{self.name} is learning {self.course}"



sahil_obj = Student()
alice_obj = Student()

# You now have empty Student objects
alice_obj.learning()
```

1. __init__ with Parameters

- You can pass multiple parameters to set up different attributes.

```python
# define a class
class Dog:

    # class attribute/ class variables / class property
    Name = "Milo"
    Color = "Brown"

    def getName(self):
        return self.Name

    def getColor(self):
        return self.Color
```

```python
class Dog:

    def __init__(self,Name,Color):
        self.Name = Name
        self.Color = Color

    def getName(self):
        return self.Name

    def getColor(self):
        return self.Color
```

```python
tommy = Dog(Name = "Tommy", Color = "Brown")
```

```python
tommy.getName()
```

```python
milo = Dog(Name = "Milo", Color = "Black")

milo.getName()
```

Explanation:

- **self**: refers to the current object (always the first parameter).
- **name** and **age**: parameters passed when creating the object.
- **self.name** and **self.age**: object attributes that store these values.

2. Default Parameters in __init__

Like normal functions, you can use default values in constructors.

```python
class Dog:
    def __init__(self, name, breed="Mixed", age=1): # class method
        self.name = name # class attributes
        self.breed = breed # class attributes
        self.age = age# class attributes

    def speak(self,color="black"): # function attributes # # class method
        return f"{self.name} is a {self.age} year old {self.breed} and have {color} color"

d1 = Dog(name = "Buddy")
d2 = Dog(name = "Max", breed = "Golden Retriever", age = 5)
#
# d1.speak(color="white")
print(d2.name, d2.breed, d2.age)
print(d2.speak())
```

#### Question 1: Define a new class called `Rectangle`. This class should have two `attributes`:

#### Question 2: Inside the `Rectangle` class, define a method called `calculate_perimeter`. This method should calculate and print the perimeter of the rectangle using the formula: Perimeter = 2 * (width + height). The method should access the width and height attributes of the Rectangle object it's called on.

#### Question 3: Create two different objects (instances) of the Rectangle class you defined in Question 1. Name them `rectangle_one` and `rectangle_two`.

#### Question 4:

* For the `rectangle_one` object:
* Change its `width` attribute to 7.5.
* Change its `height` attribute to 12.0.
* Call the calculate_perimeter method on rectangle_one.
#### Question 5:
* For the rectangle_two object:
* Leave its width and height attributes at their default values.
* Call the calculate_perimeter method on rectangle_two.

```python
class Rectangle:
    def __init__(self,width=5.0,height=10.0):
        self.width = width
        self.height = height

    def perimeter(self):
        perimeter= 2 * (self.width + self.height)
        return perimeter


rectangle1=Rectangle(width=7.5,height=12.0)
print(rectangle1.perimeter())

rectangle2=Rectangle()
print(rectangle2.perimeter())
```
