## 🐍 Python Strings


#### 📌 What is a String?


- A string in Python is a sequence of characters enclosed in single quotes (' '), double quotes (" "), or triple quotes (''' ''' or """ """).

```python
name = "Python"

city = 'Toronto'

message = "this is a single line string"


message = '''This is a
second-line string
third-line string
'''
```

### 📌 Characteristics of Strings
- Strings are immutable (cannot be changed after creation)
- Indexed (each character has a position)
- Can contain letters, numbers, spaces, and symbols

```python
# actual value
name = "sahil"

name = "sajil"

# expected representation
name = "sajil"



name[2] = 'j'

```

```python
text = "Hello"
print(text[4])
# text[0] = 'h' ❌ Not allowed
```

```python
name = "432y78352jb5j2njnjkn542nj2n4j@@@@---"
name = "432y78352jb5j2njnjkn542nj2n4j@@---"
```

### 📌 Common String Functions (Methods)
#### 1️⃣ upper() :

Converts string to uppercase.

```python
word = "python is my favorite language and it has many features with more funcations"

word_capital = word.upper()
print(word)
print(word_capital)
```

#### 2️⃣ lower() :

Converts string to lowercase.

```python
text = "PYTHON IS MY FAVORITE LANGUAGE AND IT HAS MANY FEATURES WITH MORE FUNCTIONS"
print(text.lower())
```

#### 3️⃣ strip() :

Removes leading and trailing spaces.

```python
text = "Kalam    is having a question"


print(text)

print(text.strip())
```

#### 4️⃣ replace(old, new):

Replaces part of a string.

```python
text = "I like Java"

new_text = text.replace("Java", "Python")

print(text)
print(new_text)
```

```python
name = "java java java"
```

#### 5️⃣ split()
Splits string into a list.

```python
text = "apple-banana-apple" # --> list of elements --> ["apple","banana","apple"]

text.split(" ")
# print(text.split(" "))
```

#### 6️⃣ join()
Joins list elements into a string.

```python
words = ["I", "love", "Python"] # --> I love Python


join_words = "**".join(words)
join_words
```

#### 7️⃣ find()
Finds index of substring.

```python
text = "I am Python and I am teaching Python"
print(text.find("python"))
```

#### 8️⃣ count()
Counts occurrences of a character or word.

```python
text = "banana"
print(text.count("ba"))
```

### 📌 String Length
Use len() to find string length.

```python
text = "Shloka"
len(text)
```
