### 🧐 Why Iterator Data Types in Python are so Important and Needs Good Grasp. 🧐


#### 🤘 Data Organization
* They provide versatile structures for storing and organizing collections of data, whether it's ordered sequences `(lists, tuples)` or key-value pairs `(dictionaries)`.

#### 🤘 Data Manipulation
* They offer built-in methods and operations for efficiently adding, removing, modifying, and accessing data elements.

#### 🤘 Flexibility
* `Lists` are mutable (changeable), `tuples` are immutable (unchangeable), and `dictionaries` allow for flexible key-value relationships, catering to diverse programming needs.

### Lists

#### list is a built-in dynamic-sized array (automatically grows and shrinks). We can store all types of items (including another list) in a list.
##### * list can contain duplicate elements.
##### * list can mutable, hence can modify, replace, delete items in the list.
##### * list is ordered. It maintains the order of elements based on how they are added.

```python

# list extends and adds two lists
numbers_1 = [1,2,3,4]
numbers_2 = [5,6,7,8]

# number = numbers_1 + numbers_2
# print("The new list is: ", number)

print(numbers_1)

# add the items of numbers1 to the number2 list
numbers_1.extend(numbers_2)
print("The extended list is: ", numbers_1)
```

```python
# list insert
vowels = ['a', 'e', 'i', 'u'] #
print(vowels)

alphabets = ['a','b','c']
# we now add o between 'i' and 'u'
vowels.insert(3, alphabets)
print("The vowels is: ", vowels)
```

```python
# count elements in list
numbers = [1,2,3,4,5,3,2,1]

# check the count of specific element
countTimes = numbers.count(3)
print("The count of 3 is: ", countTimes)
```

```python
# sort elements in list
numbers = [1,2,3,4,5,3,2,1]

# check the count of specific element : sorted()
numbers.sort()


# print("The sort() list is: ", numbers)
print("The sorted() list is: ", numbers)
```
