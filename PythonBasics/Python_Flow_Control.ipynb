## Conditional Statements

Conditional statements in Python are used to execute certain blocks of code based on specific conditions. These statements help control the flow of a program, making it behave differently in different situations.


### If Conditional Statement
If statement is the simplest form of a conditional statement. It executes a block of code if the given condition is true.In computer programming, the if statement is a conditional statement. It is used to execute a block of code only when a specific condition is met. For example,


### If...Else Statement

#### Suppose we need to assign different grades to students based on their scores.

- If a student scores above 90, assign grade A --> grade, score
- If a student scores above 75, assign grade B
- If a student scores above 65, assign grade C

- These conditional tasks can be achieved using the if statement.

### Python if Statement
- An `if` statement executes a block of code only when the specified condition is met.

### Indentation in Python

#### Python uses indentation to define a block of code, such as the body of an if statement.

```python
grade = ''
score = 90

if score >= 90:
    grade = 'A'
if score > 75:
    grade = 'B'

print(grade)
```

#### Here, condition is a boolean expression, such as number > 5, that evaluates to either True or False.

- If condition evaluates to True, the body of the if statement is executed.
- If condition evaluates to False, the body of the if statement will be skipped from execution.

```python
number = int(input('Enter a number: '))

# check if number is greater than 0
if number > 0:
    print("The number is a positive number")
if number < 0:
    print("The number is a negative number")


print("I checked the number")
```

### Python if…elif…else Statement
#### The `if...else` statement is used to execute a block of code among two alternatives.

#### However, if we need to make a choice between more than two alternatives, we use the `if...elif...else` statement.

```python
number = 6
number2 = 67


if number > 0:
    print('Positive number')
elif number ==6:
    print("The number is 6")
elif number < 0:
    print('Negative number')
elif number == 0:
    print("The number is zero")
else:
    print('What the hell is that number')

print('This statement is always executed')

# 4,6,14
# 4,14
```

### Python Nested if Statements
#### It is possible to include an if statement inside another if statement. For example,

```python
number = 0


# debugging --> errors


# outer if statement
if number >= 0:

    # inner if statement
    if number == 0:
      print('Number is 0')

    # inner else statement
    else:
        print('Number is positive')

# outer else statement
else:
    print('Number is negative')

# 14
```

### For Loop with Python range()
#### In Python, the range() function returns a sequence of numbers

#### The `range()` function takes up to three integer arguments:
`range(start, stop, step)`
- start (optional): The starting number of the sequence (inclusive). The default is 0 if not specified.
- stop (required): The number before which the sequence must end (exclusive). This value is never included in the generated sequence.
- step (optional): The difference between each number in the sequence. The default is 1 if not specified. This value cannot be 0

```python
list(range(10))

# range -- stop -- 0,1,2,3,4
```

### break and continue Statement
#### The `break` and `continue` statements are used to alter the flow of loops.

#### The break Statement
- The `break` statement terminates the for loop immediately before it loops through all the items.

```python
languages = ['Swift', 'Python', 'Go', 'C++']

for lang in languages:
    if lang == 'Python':
        break
    print(lang)
```

#### The continue Statement
- The `continue` statement skips the current iteration of the loop and continues with the next iteration.

```python
languages = ['Swift', 'Ruby','Python','Java' ,'Go', 'C++']

for lang in languages:
    if lang == 'Ruby':
        continue
    print(lang)
```

### Nested for loops
1. A loop can also contain another loop inside it. These loops are called nested loops.
2. In a nested loop, the inner loop is executed once for each iteration of the outer loop.

```python
# outer loop
attributes = ['Electric', 'Fast']
cars = ['Tesla', 'Porsche', 'Mercedes']

for attribute in attributes:
    for car in cars:
        print(attribute, car)
```

```python
for a,b in zip(attributes,cars):
    print(a,b)
```
