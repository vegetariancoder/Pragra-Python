### Pandas Select

- Pandas select refers to the process of extracting specific portions of data from a DataFrame.
- Data selection involves choosing specific rows and columns based on labels, positions, or conditions.
- Pandas provides various methods, such as basic indexing, slicing, boolean indexing, and querying, to efficiently extract, filter, and transform data, enabling users to focus on relevant information for analysis and decision-making.

```python
import pandas as pd
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [22, 28, 35, 19, 29],
    "City": ["New York", "Chicago", "New York", "Boston", "Chicago"],
    "Stream": ["Math", "Science", "Math", "Arts", "Science"],
    "Score": [85, 92, 78, 95, 88],
    "Sales_Q1": [10, 15, 8, 12, 14],
    "Sales_Q2": [12, 18, 9, 11, 15],
}

df = pd.DataFrame(data)

df
```

1. Basic Selection Using .loc[]

The first argument of .loc[] filters rows based on a boolean condition, while the second argument selects the columns you want to keep

```python
result_basic = df.loc[df["Age"] > 25, ["Name", "City"]]
```

2. Common Row Filtering Conditions

Comparison Operators: Use >, <, ==, !=, >=, <=.Membership (isin) Select rows where values match any item in a list.

```python
df.loc[df['Stream'].isin(['Math', 'Science']), ['Name']]
```

Multiple Conditions: Combine conditions using & (AND), | (OR), and ~ (NOT). Always use parentheses around each condition.

```python
df.loc[(df['Age'] > 20) & (df['Score'] < 90), :]
```

3. Dynamic Column Selection

You can also select columns based on conditions, such as their names or the data they contain
- By Name Pattern: Use the filter() method to select columns that match a substring or regex.

```python
# By Name Pattern
df.filter(like='Sales')
```

4. Alternative MethodsBoolean Indexing

- For row filtering only, you can use df[df['Age'] > 25].
- Query Method: Use the query() method for a cleaner, SQL-like string syntax.

```python
df.query('Age > 25 & City == "New York"')[['Name', 'Score']]
```
