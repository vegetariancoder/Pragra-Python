### Import Library

```python
import pandas as pd
import pymysql
from sqlalchemy import create_engine,text
```

### Create Connection

```python
# Database connection
username = "root"
password = "password"
host = "localhost"
database = "leetcode"

engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}/{database}"
)



# SQL queries
create_table_query = """
CREATE TABLE IF NOT EXISTS pragra_table_info (
    id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(100),
    age INT,
    course VARCHAR(100)
)
"""

insert_query = """
INSERT INTO pragra_table_info (name, age, course)
VALUES
    ('Alice', 20, 'Python'),
    ('Bob', 22, 'Data Science'),
    ('Charlie', 19, 'Machine Learning'),
    ('David', 21, 'Web Development'),
    ('Emma', 23, 'Cloud Computing')
"""

# Execute queries
with engine.connect() as conn:

    # Create table
    conn.execute(text(create_table_query))

    # Insert rows
    conn.execute(text(insert_query))

    # Commit changes
    conn.commit()

    print("Table created and 5 rows inserted successfully!")
```

```python
# Read table using pandas
df = pd.read_sql("SELECT * FROM students", engine)

# read_sql --> sql query, connection

# Display dataframe
df
```
