### Import Library

```python
import pandas as pd
from sqlalchemy import create_engine,text
```

### Create Connection

```python
# Database connection
username = "root"
password = "password"
host = "localhost"
database = "leetcode"

import urllib.parse

# 1. URL-encode only the password

print("Connecting to the database...")

engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}/{database}"
)

print("Database connection established!")



# # SQL queries
create_table_query = """
CREATE TABLE IF NOT EXISTS pragra_sample_table_2 (
    id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(100),
    age INT,
    course VARCHAR(100)
)
"""
# #
insert_query = """
INSERT INTO pragra_sample_table_2 (name, age, course)
VALUES
    ('Alice', 20, 'Python'),
    ('Bob', 22, 'Data Science'),
    ('Charlie', 19, 'Machine Learning'),
    ('David', 21, 'Web Development'),
    ('Emma', 23, 'Cloud Computing')
"""
# #
# #
# #
# # # Execute queries
with engine.connect() as mycon:

    # Create table

    print("Table creation and insertion started...")

    mycon.execute(text(create_table_query))

    # Insert rows
    mycon.execute(text(insert_query))

    print("Table created and 5 rows inserted successfully!")

    # Commit changes
    mycon.commit()

    print("Table Commited created and 5 rows inserted successfully!")
```

### Reading the table and Creating DataFrame

```python
# Read table using pandas
df = pd.read_sql(" SELECT * FROM pragra_sample_table_2 ", engine)

# Display dataframe
df
```

### Writing the DataFrame as Table

```python
df.to_sql(
    name='pragra_write_tabl2',    # Name of the SQL table
    con=engine,            # Database connection engine
    if_exists='append',    # What to do if the table already exists. # append / # overwrite
    index=False            # Do not write the DataFrame index as a column
)
```
