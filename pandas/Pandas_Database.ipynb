### Import Library

```python
import pandas as pd
from sqlalchemy import create_engine,text
```

### Create Connection

```python
# Replace with your own MySQL credentials
username = "root"
password = "password"
host = "localhost"   # or IP of your MySQL server
database = "leetcode"


def connectDatabase():
    """
    Establishes a connection to a MySQL database and prints the database version.

    This function initializes a global SQLAlchemy engine variable and
    connects to the database using the given credentials, executing a query
    to retrieve and display the database version.

    :raises sqlalchemy.exc.OperationalError: If there is an issue connecting to the
        database.
    :return: None
    """

    # created global param : engine
    global engine
    engine = create_engine(f"mysql+pymysql://{username}:{password}@{host}/{database}")
    print(engine)

    # making connection with database
    with engine.connect() as conn:
        print(conn.execute(text("SELECT VERSION()")).scalar())


# Create SQLAlchemy engine
connectDatabase()
```
