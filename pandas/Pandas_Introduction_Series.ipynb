### Pandas Series
- A Pandas Series is a one-dimensional labeled array-like object that can hold data of any type.
- A Pandas Series can be thought of as a column in a spreadsheet or a single column of a DataFrame.
- The labels associated with a Series are called its index.

```python
# labels  data
# 0    'John'
# 1    30
# 2    6.2
# 3    False
# dtype: object
```

```python
import pandas as pd
```

The labels are the index values assigned to each data point, while the data represents the actual values stored in the Series.

### Create a Pandas Series

```python
import pandas as pd

# create a list
data = [10, "sahil", 30, 40, "Andy", 24.9]

# create a series from the list
my_series = pd.Series(data)


my_series
```

### Labels
- The labels in the Pandas Series are index numbers by default. Like in dataframe and array, the index number in series starts from 0.

```python
import pandas as pd

# create a list
data = [10, 20, 30, 40, 50]

# create a series from the list
my_series = pd.Series(data)

# display third value in the series
my_series
```

```python
# We can also specify labels while creating the series using the index argument in the Series() method.

import pandas as pd

# create a list
myList = [1, 3, 5, 6]


# create a series and specify labels
my_series = pd.Series(myList, index = ["One", "Two", "Three", "Four"])
my_series
```

### Create Series From a Python Dictionary

```python
import pandas as pd

# create a dictionary
grades = {"Semester1": 3.25, "Semester2": 3.28, "Semester3": 3.75,"Semester4": 3.75,"Semester5":4.05}


# create a series from the dictionary
my_series = pd.Series(grades)

# display the series
type(my_series)
```

```python
# further customize the series by selecting specific items from the dictionary using the index argument

grades = {1: 3.25, 3: 3.28, 2: 3.75, 4: 3.75}


# create a series from the dictionary
my_series = pd.Series(grades)

my_series
```

### Some Important Pandas Series Methods

# Pandas Series Methods Reference

A quick-reference guide for common exploratory and manipulation methods used on a pandas `Series`.

---

### 1. `head()` and `tail()`
Used to quickly inspect the first or last few rows of a Series.
* **`Series.head(n)`**: Returns the first `n` rows (defaults to 5).
* **`Series.tail(n)`**: Returns the last `n` rows (defaults to 5).

```python

# Get the top 2 value from the Series
my_series.head(7) # n is the number of rows you want to see

# Get the bottom 2 values from the Series
# my_series.tail(2)
```

### 2. Statistical Aggregations

#### Quick scalar summaries of the numerical data within the Series.


1. Series.mean(): Computes the average value.
2. Series.sum(): Adds up all the values.
3. Series.max(): Finds the highest value.
4. Series.min(): Finds the lowest value.

```python
mean_val = my_series.mean()

sum_val  = my_series.sum()

max_val  = my_series.max()

min_val  = my_series.min()
```

### 3. Frequency Distribution
1. Series.value_counts(): Returns a Series containing counts of unique values, sorted in descending order of frequency.

```python
my_series.value_counts()
```

### 4. Cardinality and Uniqueness

1. Series.unique(): Returns an array of the unique values in the order they appear.
2. Series.nunique(): Returns the number of unique values (excludes NaNs by default).

```python
unique_elements = my_series.unique()
num_of_uniques   = my_series.nunique()
```

### 5. Sorting Data
1. Series.sort_values(ascending=True): Sorts the Series by its actual data values.
2. Series.sort_index(ascending=True): Sorts the Series by its index labels.

```python
my_series
```

### 6. Applying Functions

1. Series.apply(func): Invokes a function on each element of the Series. Ideal for element-wise transformations using custom or lambda functions.

```python
def double(n):
    return n * 2
```

```python
# Example: Double every value in the series using a lambda function
my_series.apply(double)
```

🧪 Practice Question 2: Applying Conditions and Filters
Task:
You have a Series of exam scores:

`scores = pd.Series([75, 82, 91, 68, 87, 90], index=['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'])`


Instructions:

- Print the names of students who scored more than 80.
- Calculate and print the number of students who passed (score >= 70).
- Use .apply() to convert scores to a grade (A for >=90, B for 80-89, etc.)

```python
scores = pd.Series([75, 82, 91, 68, 87, 90], index=['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'])

def changeScoreToGrade(score):
    if score>=90:
        return 'A'
    elif 80 <= score < 90:
        return 'B'
    else:
        return 'C'

scores.apply(changeScoreToGrade)




```
