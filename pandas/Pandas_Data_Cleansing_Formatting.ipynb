## Pandas Data Cleaning

### Data cleaning means fixing and organizing messy data. Pandas offers a wide range of tools and functions to help us clean and preprocess our data effectively.

## Data cleaning often involves:

1. Dropping irrelevant columns.
2. Renaming column names to meaningful names.
3. Making data values consistent.
4. Replacing or filling in missing values.

## Drop Rows With Missing Values
- In Pandas, we can drop rows with missing values using the dropna() function. For example,

```python
#name,age,salary
#1,'sahil,1200
#manish,20,1300
#ranjani,,1200

# column data type != row data type (dataset data type)

# 1 --> look into data
# 2 --> EDA (Exploratory Data Analysis) --> sanity checkings on data (based on columns and rows) --flag
```

```python
import pandas as pd

# define a dictionary with sample data which includes some missing values
data = {
    'A': [1, 2, 3, None, 5],
    'B': [None, 2, 3, 4, 5],
    'C': [1, 2, None, None, 5]
}

df = pd.DataFrame(data)

df

# sanity
# Check the missing values in each column
# Data Type (row and col)
# Duplicates
# count of rows in each column
# number of columns


# use dropna() to remove rows with any missing values
df_cleaned = df.dropna()

df_cleaned
```

## Fill Missing Values

- To fill the missing values in Pandas, we use the fillna() function.



```python
import pandas as pd

# define a dictionary with sample data which includes some missing values
data = {
    'A': [1, 2, 3, None, 5],
    'B': [None, 2, 3, 4, 5],
    'C': [1, 2, None, None, 5]
}

# my_list = [1, 2, 3, 2.8, 5] # mean (avg)
# Output = 1,5


df = pd.DataFrame(data)
#
# # filling NaN values with 0
df.fillna(0, inplace=True)

df

```

## Use Aggregate Functions to Fill Missing Values

### Instead of filling with 0, we can also use aggregate functions to fill missing values.

```python
import pandas as pd

# define a dictionary with sample data which includes some missing values
data = {
    'A': [1, 2, 3, None, 5],
    'B': [None, 2, 3, 4, 5],
    'C': [1, 2, None, None, 5]
}

df = pd.DataFrame(data)

# measurement of central tendency -- mean, median, mode


# filling NaN values with the mean of each column
df.fillna(df.median(), inplace=True)

df
```

# 🛍️ Retail Orders Data Analysis

You are provided with a dataset containing online order records from a retail store. Your goal is to clean the data and extract insights through basic aggregations.

---

## 📁 Dataset Columns

| Column Name     | Description                             |
|-----------------|-----------------------------------------|
| OrderID         | Unique identifier for each order        |
| CustomerName    | Name of the customer                    |
| City            | City where the order was placed         |
| OrderAmount     | Total amount of the order               |
| PaymentMethod   | Method used for payment (Credit/PayPal) |
| OrderDate       | Date the order was placed               |

---

## ✅ Instructions

### 🧹 Data Cleansing
- Remove rows where either `OrderID` or `CustomerName` is missing.
- Fill missing values in the `City` column with `'Unknown'`.

---

### 📊 Aggregations
- Calculate the total `OrderAmount` per `City`.
- Find the average `OrderAmount` per `CustomerName`.
- Count the number of orders for each `PaymentMethod`.
- Identify the customer who made the highest number of orders.

---

### 🎁 Bonus (Optional)
- Group the data by `City` and `PaymentMethod`, and calculate:
  - Average `OrderAmount`
  - Total `OrderAmount`


```python
data = {
    "OrderID": [101, 102, 103, 104, None, 106, 107, 108, 109, 110, 111],
    "CustomerName": ["Alice", "Bob", "Charlie", "Alice", "David", "Eve", "Alice", "Bob", "Frank", "Grace", "Alice"],
    "City": ["New York", None, "Chicago", "New York", "Boston", "Chicago", "New York", "Chicago", "Boston", None, "New York"],
    "OrderAmount": [250.00, 300.50, None, 400.00, 150.00, 220.00, 180.75, 340.25, 199.99, 310.00, 120.00],
    "PaymentMethod": ["Credit Card", "PayPal", "Credit Card", "Debit Card", "Credit Card", "PayPal", "Credit Card", "PayPal", "Credit Card", "Debit Card", "Credit Card"],
    "OrderDate": pd.to_datetime(["2023-08-01", "2023-08-01", "2023-08-02", "2023-08-03", "2023-08-04",
                                 "2023-08-05", "2023-08-05", "2023-08-06", "2023-08-07", "2023-08-08", "2023-08-08"])
}

df = pd.DataFrame(data)

df
```

```python
df_cleaned = df.dropna(subset=['OrderID', 'CustomerName'])
df_cleaned['City'] = df_cleaned['City'].fillna('Unknown')
total_amount_per_city = df_cleaned.groupby('City')['OrderAmount'].sum()
total_amount_per_city
```

```python
avg_amount_per_customer = df_cleaned.groupby('CustomerName')['OrderAmount'].mean()
avg_amount_per_customer
```

```python
orders_per_payment = df_cleaned['PaymentMethod'].value_counts()
orders_per_payment

# idxmax -->
```

```python
orders_per_customer = df_cleaned['CustomerName'].value_counts()
top_customer = orders_per_customer.idxmax()
print(top_customer,orders_per_customer[top_customer])
```
