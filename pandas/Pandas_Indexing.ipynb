### Pandas Dataframe Index
- Index in pandas dataframe act as reference for each row in dataset.
- It can be numeric or based on specific column values.
- The default index is usually a RangeIndex starting from 0, but you can customize it for better data understanding.

```python
#     Name    Age      City
# 0   John     25  New York
# 1  Alice     28    London
# 2    Bob     32     Paris
```

### Create Indexes in Pandas
#### Pandas offers several ways to create indexes. Some common methods are as follows:

1. Default Index
2. Setting Index
3. Creating a Range Index

### Default Index

- When we create a DataFrame or Series without specifying an index explicitly, Pandas assigns a default integer index starting from 0. For example,

```python
import pandas as pd

data = {'Name': ['John', 'Alice', 'Bob'],
        'Age': [25, 28, 32],
        'City': ['New York', 'London', 'Paris']}



df = pd.DataFrame(data)
df
```

### Setting Index

- We can set an existing column as the index using the `set_index()` method. For example,

```python
import pandas as pd

# create dataframe
data = {'Name': ['John', 'Alice', 'Bob'],
        'Age': [25, 28, 32],
        'City': ['New York', 'London', 'Paris']}

df = pd.DataFrame(data)


# set the 'Age' column as index
df.set_index('Age',inplace=True)

df
```

```python
import pandas as pd

# create dataframe
data = {'Name': ['John', 'Alice', 'Bob', 'Niharika'],
        'Age': [25, 28, 32, 25],
        'City': ['New York', 'London', 'Paris', 'Cuttack']}

df = pd.DataFrame(data)


# set the 'Name' column as index
df.set_index('Age', inplace=True)

df
```

### Creating a Range Index

- We can create a range index with specific start and end values using the `RangeIndex()` function. For example,

```python
import pandas as pd

# create dataframe
data = {'Name': ['John', 'Alice', 'Bob'],
        'Phone_Number': [25, 25, 32],
        'Phone_Number_2': [25, 25, 32]
        }



# df = pd.DataFrame(data)

# create a range index
df = pd.DataFrame(data, index=pd.RangeIndex(101,104, name='IndexCol')) # startvalue , endvalue, name

df
```

### Modifying Indexes in Pandas
- Pandas allows us to make changes to indexes easily. Some common modification operations are:

1. Renaming Index
2. Resetting Index

### Renaming Index
- We can rename an index using the `rename()` method.

```python
import pandas as pd

# create a dataframe
data = {'Name': ['John', 'Alice', 'Bob'],
        'Age': [25, 28, 32],
        'City': ['New York', 'London', 'Paris']}


df = pd.DataFrame(data)



# rename index
renameIndex = {0: '0_index', 1: 'first_index', 2: 'second_index'}
df.rename(index=renameIndex, inplace=True)

# display dataframe after index is renamed
# print('Modified DataFrame')
df
```

### Resetting Index

- We can reset the index to the default integer index using the reset_index() method.

```python
import pandas as pd

data = {'Name': ['John', 'Alice', 'Bob'],
        'Age': [25, 28, 32],
        'City': ['New York', 'London', 'Paris']}

# create a dataframe
df = pd.DataFrame(data)

# rename index
df.rename(index={0: 'A', 1: 'B', 2: 'C'}, inplace=True)


# display dataframe

# reset index
df.reset_index(inplace=True)

# display dataframe after index is reset

df
```

### Access Rows by Index
- We can access rows of a DataFrame using the .iloc property.

```python
import pandas as pd

# creating a sample dataframe
data = pd.DataFrame({'Brand': ['Maruti', 'Hyundai', 'Tata',
                               'Mahindra', 'Maruti', 'Hyundai',
                               'Renault', 'Tata', 'Maruti'],
                     'Year': [2012, 2014, 2011, 2015, 2012,
                              2016, 2014, 2018, 2019],
                     'Kms Driven': [50000, 30000, 60000,
                                    25000, 10000, 46000,
                                    31000, 15000, 12000],
                     'City': ['Gurgaon', 'Delhi', 'Mumbai',
                              'Delhi', 'Mumbai', 'Delhi',
                              'Mumbai', 'Chennai',  'Ghaziabad'],
                     'Mileage':  [28, 27, 25, 26, 28,
                                  29, 24, 21, 24]})

data
```

#### Python loc() function

```python
data.loc[2:7,["Brand","Year","City"]]
```

```python
# selecting cars with brand 'Maruti' and Mileage > 25
display(
    data.loc[(data.Brand == 'Maruti') & (data.Mileage > 25) & (data.City == 'Mumbai')]
)
```

```python
data.loc[(data.Brand == 'Maruti') & (data.Mileage > 25)]
```
