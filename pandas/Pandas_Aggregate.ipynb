## What is Aggregation
#### Aggregation = combining multiple values into one summary value (like mean, sum, count).
#### It’s typically used with:
- groupby()
- agg()

## Pandas Aggregate Function

### Aggregate function in Pandas performs summary computations on data, often on grouped data. But it can also be used on Series objects.

### This can be really useful for tasks such as calculating mean, sum, count, and other statistics for different groups within our data.

#### Syntax
Here's the basic syntax of the aggregate function,

`df.aggregate(func, *args, **kwargs)`

- `func` - an aggregate function like sum, mean, etc.
- `*args and **kwargs` - additional arguments that can be passed to the aggregation functions.



```python
import pandas as pd

data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B'],
    'Region': ['North', 'North', 'South', 'South', 'North', 'South'],
    'Price': [10, 15, 20, 25, 30, 35],
    'Discount': [5, 10, 7, 15, 5, 5]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# dataframe.groupby("column on which to group (distinct)")["column(s) to aggregate"].aggregationMethod()
```

#### Grouped Aggregation
- Use .groupby() to aggregate based on one or more keys.

```python
# df.groupby("Region")["Price"].sum()
df.groupby(["Region", "Category"])["Price"].mean()
```

#### Multiple Aggregations on Same Column
- Use .agg() with a list of functions:

```python
df.groupby("Region")["Price"].agg(["sum", "mean", "count", "median","min","max"])
```

#### Different Aggregations for Different Columns

```python
agg_dict = {
    "Price": ["sum", "mean"],
    "Discount": "min"
}

df.groupby("Region").agg(agg_dict)
```

#### Aggregating without GroupBy — DataFrame.agg()

```python
agg_dict = {
    "Price": ["mean", "sum", "min"],
    "Discount": ["max", "min","mean", "sum"]
}

df.agg(agg_dict)
```
